```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)

# Sample size
n = 1000

# 1. Normal Distribution
mean, std = 0, 1
normal_samples = stats.norm.rvs(loc=mean, scale=std, size=n)

# 2. Binomial Distribution
n_trials, p = 10, 0.5
binomial_samples = stats.binom.rvs(n=n_trials, p=p, size=n)

# 3. Poisson Distribution
lambda_param = 3
poisson_samples = stats.poisson.rvs(mu=lambda_param, size=n)

# 4. Exponential Distribution
beta = 1.0
exponential_samples = stats.expon.rvs(scale=beta, size=n)

# 5. Uniform Distribution
a, b = 0, 1
uniform_samples = stats.uniform.rvs(loc=a, scale=b, size=n)

# 6. Student's t Distribution
df = 5
t_samples = stats.t.rvs(df=df, size=n)

# Create subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Common Theoretical Distributions')

# Plot distributions
sns.histplot(normal_samples, kde=True, ax=axes[0,0])
axes[0,0].set_title('Normal Distribution')

sns.histplot(binomial_samples, kde=True, ax=axes[0,1])
axes[0,1].set_title('Binomial Distribution')

sns.histplot(poisson_samples, kde=True, ax=axes[0,2])
axes[0,2].set_title('Poisson Distribution')

sns.histplot(exponential_samples, kde=True, ax=axes[1,0])
axes[1,0].set_title('Exponential Distribution')

sns.histplot(uniform_samples, kde=True, ax=axes[1,1])
axes[1,1].set_title('Uniform Distribution')

sns.histplot(t_samples, kde=True, ax=axes[1,2])
axes[1,2].set_title("Student's t Distribution")

plt.tight_layout()
plt.show()

# Quick reference for generating samples:
"""
Normal: stats.norm.rvs(loc=mean, scale=std, size=n)
Binomial: stats.binom.rvs(n=n_trials, p=p, size=n)
Poisson: stats.poisson.rvs(mu=lambda_param, size=n)
Exponential: stats.expon.rvs(scale=beta, size=n)
Uniform: stats.uniform.rvs(loc=a, scale=b, size=n)
Student's t: stats.t.rvs(df=df, size=n)
"""
```
